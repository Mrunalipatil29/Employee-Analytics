## visualization

Salary Distribution Chart

```python
plt.figure(figsize=(8,5))

plt.hist(df["Monthly_Salary"], bins=20)

plt.xlabel("Monthly_Salary")
plt.ylabel("Number of Employees")
plt.title("Monthly Salary Distribution")

plt.show()
```

Department-wise Average Salary Chart

```python
avg_salary = df.groupby("Department")["Monthly_Salary"].mean().sort_values()

avg_salary.plot(kind="barh", figsize=(8,5))

plt.xlabel("Average Monthly Salary")
plt.ylabel("Department")
plt.title("Average Salary by Department")

plt.show()
```

```python
avg_performance = df.groupby("Department")["Performance_Score"].mean().sort_values(ascending=False)

avg_performance
```

Performance by Department Chart


```python
sns.barplot(
    x=avg_performance.index,
    y=avg_performance.values
)

plt.xlabel("Department")
plt.ylabel("Average Performance Score")
plt.title("Average Performance by Department")

plt.xticks(rotation=45)

plt.show()
```

Top 10 Employees by Performance

```python
top_performance = df.nlargest(10, "Performance_Score")[["Employee ID",
    "Department",
    "Job_Title",
    "Performance_Score"
]]

top_performance
```

Performance Score Distribution

```python
plt.figure(figsize=(8,5))

plt.hist(df["Performance_Score"], bins=10)

plt.xlabel("Performance_Score")
plt.ylabel("Number of Employees")
plt.title("Performance Score Distribution")

plt.show()
```
